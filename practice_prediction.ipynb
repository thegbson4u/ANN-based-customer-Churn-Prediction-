```python
model=load_model('model.h5')
```

```python
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender = pickle.load(file)
```

```python
with open('one_hot_encoder_ge.pkl','rb') as file:
     one_hot_encoder = pickle.load(file)
```

```python
with open("scalar.pkl",'rb') as file:
    scalar= pickle.load(file)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Female',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
geography = one_hot_encoder.transform([[input_data["Geography"]]]).toarray()
geography_df = pd.DataFrame(geography,columns=one_hot_encoder.get_feature_names_out(["Geography"]))
geography_df
```

```python
input_data = pd.DataFrame([input_data])
input_data
```

```python
input_data["Gender"]=label_encoder_gender.transform(input_data["Gender"])
```

```python
input_data=pd.concat([input_data.drop("Geography",axis=1),geography_df],axis=1)
```

```python
input_data
```

```python
input_scaled= scalar.transform(input_data)
input_scaled
```

```python
prediction = model.predict(input_scaled)
result = prediction[0][0]
```

```python
model_predicted=1 if result >0.6 else 0 
model_predicted
```
