```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,StandardScaler
import pickle
from sklearn.model_selection import train_test_split
import datetime
```

```python
data=pd.read_csv("Churn_Modelling.csv")
data.head()
```

```python
data=data.drop(["RowNumber","CustomerId","Surname"],axis=1)
```

```python
one_hot_encoder_ge = OneHotEncoder()
encoder_ge = one_hot_encoder_ge.fit_transform(data[["Geography"]]).toarray()
encoder_ge
```

```python
one_hot_encoder_ge.get_feature_names_out()
```

```python
geo_encoded_df= pd.DataFrame(encoder_ge,columns=one_hot_encoder_ge.get_feature_names_out())
geo_encoded_df.head()
```

```python
data=pd.concat([data.drop(["Geography"],axis=1),geo_encoded_df],axis=1)
```

```python
data.head()
```

```python
encoder_ge =LabelEncoder()
data["Gender"]=encoder_ge.fit_transform(data["Gender"])
data.tail()
```

```python
with open("enocder_ge.pkl","wb")as file:
    pickle.dump(encoder_ge,file)
```

```python
with open("one_hot_enocder_ge.pkl","wb") as file:
    pickle.dump(one_hot_encoder_ge,file)
```

```python
#existed is target prediction
```

```python
x=data.drop(["Exited"],axis=1)
x
```

```python
y=data["Exited"]
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=98)
```

```python
x_train.head()
```

```python
scaler=StandardScaler()
```

```python
x_train= scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
x_train
```

```python
with open ("scaler.pkl","wb") as file:
    pickle.dump(scaler,file)
```

### ANN IMPLENTATION

```python
import tensorflow as tf

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
```

## Build ANN Model

```python
model = Sequential([
    Input(shape=(x_train.shape[1],)),
    Dense(64,activation="relu"),
    Dense(32,activation="relu"),
    Dense(1,activation="sigmoid")
]
                  )
```

```python
model.summary()
```

```python
opt = tf.keras.optimizers.Adam(learning_rate=0.1)
loss = tf.keras.losses.BinaryCrossentropy()
loss
```

## compile model

```python
model.compile(optimizer=opt,loss="binary_crossentropy",metrics=["accuracy"])
```

#### setup tensorboards

```python
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
log_dir = "logs/fit/" +datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback=TensorBoard(log_dir= log_dir,histogram_freq=1)
```

#### early stopping 

```python
early_stopping_callbacks = EarlyStopping(monitor='val_loss',patience=3,restore_best_weights=True)
```

```python
history=model.fit(x_train,y_train,validation_data=(x_test,y_test),epochs=150,callbacks = [tensorflow_callback,early_stopping_callbacks])

```

```python
model.save('model.h5')
```
